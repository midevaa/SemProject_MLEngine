```python
import pandas as pd
import numpy as np

# Load the data in FloraDB from Hugging Face
df = pd.read_csv("hf://datasets/Ichlibitiche/floradb-houseplants-care-sample/floradb_sample.csv")

print("Dataset loaded successfully!")
print(f"Shape: {df.shape}")
```

**Question a:** How many rows and columns are contained in the dataset?

```python
rows, cols = df.shape
print(f"Number of rows: {rows}")
print(f"Number of columns: {cols}")
```

**Question b:** What datatypes are contained in the dataset?

```python
print("Data types in the dataset:")
print(df.dtypes)
```

**Question c:** Is the dataset complete? i.e. no missing values?

```python
missing = df.isnull().sum()
print("Missing values per column:")
print(missing[missing > 0])
print(f"\nTotal missing values: {missing.sum()}")
print(f"Columns with no missing values: {(missing == 0).sum()}")
```

**Question d:** Slice out the first 15 rows and last 20 rows and merge them into a new dataframe titled `df_sample`. Display this new dataframe.

```python
first_15 = df.head(15)
last_20 = df.tail(20)

df_sample = pd.concat([first_15, last_20], ignore_index=True)

print(f"df_sample shape: {df_sample.shape}")
print("\ndf_sample preview:")
display(df_sample)
```
